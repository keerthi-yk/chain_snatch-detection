```python
!pip install ultralytics opencv-python

import cv2
import numpy as np
from ultralytics import YOLO
from collections import defaultdict
```

```python
model = YOLO("yolov8n.pt")  # lightweight & fast
```

```python
video_path = "/kaggle/input/datasets/aniketsahu09/chain-snatching-cctv-dataset/Chain_Snatching_Videos/Chain_Snatching169.mp4" 

cap = cv2.VideoCapture(video_path)

width = int(cap.get(3))
height = int(cap.get(4))
fps = int(cap.get(cv2.CAP_PROP_FPS))

out = cv2.VideoWriter("output.mp4",
                      cv2.VideoWriter_fourcc(*'mp4v'),
                      fps, (width, height))
```

```python
trajectories = defaultdict(list)
speeds = defaultdict(list)
labels = {}
```

```python
frame_id = 0
final_results = []   # for CSV

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame, device="cpu")

    boxes = results[0].boxes.xyxy.cpu().numpy()
    classes = results[0].boxes.cls.cpu().numpy()

    current_positions = {}

    for i, box in enumerate(boxes):
        x1, y1, x2, y2 = map(int, box)
        cls = int(classes[i])

        # Track only person (0) and bike (3)
        if cls not in [0, 3]:
            continue

        cx = (x1 + x2) // 2
        cy = (y1 + y2) // 2

        obj_id = i

        trajectories[obj_id].append((cx, cy))
        current_positions[obj_id] = (cx, cy)

        # Speed calculation
        if len(trajectories[obj_id]) > 1:
            prev = trajectories[obj_id][-2]
            dist = np.linalg.norm(np.array((cx, cy)) - np.array(prev))
            speeds[obj_id].append(dist)

    # 🎯 CLASSIFICATION
    for obj_id in trajectories:
        if len(speeds[obj_id]) < 6:
            continue

        current_speed = np.mean(speeds[obj_id][-3:])
        prev_speed = np.mean(speeds[obj_id][-6:-3])

        # ✅ ACCELERATION
        acceleration = current_speed - prev_speed

        # 🔹 ABSOLUTE SPEED LEVEL
        if current_speed > 100:
            speed_level = "High"
        elif current_speed > 40:
            speed_level = "Medium"
        else:
            speed_level = "Low"

        # 🔹 CHANGE
        if acceleration > 5:
            change = "Increased"
        elif acceleration < -5:
            change = "Decreased"
        else:
            change = "Stable"

        speed_label = f"{change} ({speed_level})"

        # 🔹 DIRECTION CHANGE
        if len(trajectories[obj_id]) > 5:
            p1 = np.array(trajectories[obj_id][-5])
            p2 = np.array(trajectories[obj_id][-1])
            direction_change = np.linalg.norm(p2 - p1)
        else:
            direction_change = 0

        # 🔹 PATTERN
        if change == "Increased" and current_speed > 80:
            pattern = "Rapid Acceleration"
        elif change == "Decreased":
            pattern = "Sudden Slowdown"
        elif current_speed < 10:
            pattern = "Very Slow Movement"
        else:
            pattern = "Normal Movement"

        # 🔹 EVENT
        if pattern in ["Rapid Acceleration", "Sudden Slowdown"] or direction_change > 30:
            event = "Suspicious Movement"
        else:
            event = "Normal"

        labels[obj_id] = (speed_label, pattern, event)

        # ✅ SAVE FOR CSV
        final_results.append({
            "frame": frame_id,
            "object_id": obj_id,
            "current_speed": float(current_speed),
            "previous_speed": float(prev_speed),
            "acceleration": float(acceleration),
            "pattern": pattern,
            "event": event
        })

    # 🎥 DRAW LABELS (LEFT SIDE)
    y_offset = 30

    for idx, obj_id in enumerate(labels):
        speed_label, pattern, event = labels[obj_id]

        text = f"F:{frame_id} ID:{obj_id} | {speed_label} | {pattern} | {event}"

        color = (0, 0, 255) if event == "Suspicious Movement" else (0, 255, 0)

        cv2.putText(frame, text,
                    (10, y_offset + idx * 25),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6, color, 2)

    out.write(frame)
    frame_id += 1

cap.release()
out.release()
```

```python
import pandas as pd

df = pd.DataFrame(final_results)
df.to_csv("motion_analysis_results.csv", index=False)

print("CSV SAVED SUCCESSFULLY")
```
